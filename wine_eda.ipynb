```python
columns=df.columns
(fig, ax)=plt.subplots(4,4,figsize=(20,20))
for i, column in enumerate(columns):
  ax = ax.flatten()
  sns.kdeplot(
      data=df,
      x=column,
      hue= df.quality,
      ax=ax[i]
  )
  ax[i].set_title(f"{column} distribution")
  ax[i].set_xlabel(None)
for i in range(i+1, len(ax)):
  ax[i].axis("off")
```
